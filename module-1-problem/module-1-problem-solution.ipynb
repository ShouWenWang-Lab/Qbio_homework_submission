## GFP Fluorescence Time-Series Analysis

This notebook loads the GFP fluorescence time-series data, creates a publication-quality plot, calculates the requested metrics, and briefly interprets the results in the context of gene expression.

## 1. Load the data using NumPy

Place `fluorescence_data.csv` in the same folder as this notebook.

```python
import numpy as np
import matplotlib.pyplot as plt

# Load the CSV file using NumPy.
# The CSV is expected to have two columns:
# Time (minutes), Fluorescence (A.U.)
data = np.genfromtxt(
    'fluorescence_data.csv',
    delimiter=',',
    names=True,
    dtype=float,
    encoding='utf-8'
)

time = data['Time (minutes)']
fluorescence = data['Fluorescence (A.U.)']

print(f'Number of measurements: {len(time)}')
print(f'Time range: {time.min():.1f} to {time.max():.1f} minutes')
print(f'Fluorescence range: {fluorescence.min():.2f} to {fluorescence.max():.2f} A.U.')
```

## 2. Plot fluorescence versus time

The plot shows how GFP fluorescence changes after induction. Clear axis labels, units, and an informative title are included.

```python
plt.figure(figsize=(8, 5))
plt.plot(time, fluorescence, marker='o', linewidth=2, markersize=5)
plt.xlabel('Time (minutes)', fontsize=12)
plt.ylabel('Fluorescence (A.U.)', fontsize=12)
plt.title('GFP Fluorescence After Induction in E. coli', fontsize=14)
plt.grid(True, alpha=0.3)
plt.tight_layout()
plt.show()
```

## 3. Calculate the requested metrics

### Maximum fluorescence
The maximum fluorescence is the highest measured GFP fluorescence level.

### Time of maximum fluorescence
This is the time point corresponding to the maximum fluorescence measurement.

### Average rate of fluorescence increase from 10 to 60 minutes
The average rate is calculated as:

$$\text{Average rate} = \frac{F_{60}-F_{10}}{60-10}$$

where fluorescence is measured in A.U. and time in minutes.

```python
# Maximum fluorescence and its time
max_index = np.argmax(fluorescence)
max_fluorescence = fluorescence[max_index]
time_of_max = time[max_index]

# Find the measurements at t = 10 and t = 60 minutes.
# This assumes the dataset contains these exact time points.
t10_index = np.where(np.isclose(time, 10))[0]
t60_index = np.where(np.isclose(time, 60))[0]

if len(t10_index) == 0 or len(t60_index) == 0:
    raise ValueError('The dataset must contain measurements at t=10 and t=60 minutes.')

f10 = fluorescence[t10_index[0]]
f60 = fluorescence[t60_index[0]]

average_rate = (f60 - f10) / (60 - 10)

print(f'Maximum fluorescence level: {max_fluorescence:.2f} A.U.')
print(f'Time at maximum fluorescence: {time_of_max:.1f} minutes')
print(f'Fluorescence at 10 minutes: {f10:.2f} A.U.')
print(f'Fluorescence at 60 minutes: {f60:.2f} A.U.')
print(f'Average rate of fluorescence increase (10–60 min): {average_rate:.4f} A.U./minute')
```

## 4. Interpretation

The plot shows the change in GFP fluorescence over time following induction of gene expression in *E. coli*. An increase in fluorescence indicates increasing GFP production, so the fluorescence trajectory provides a proxy for the accumulation of GFP protein in the cells. The maximum fluorescence represents the highest measured level of GFP expression during the experiment, while the time of the maximum indicates when this highest measured level was reached. The average rate between 10 and 60 minutes describes how quickly fluorescence increased over this interval; a positive value indicates net GFP accumulation during this period.

## 5. Conclusion

Overall, the time-series analysis provides a quantitative view of GFP expression after induction. The fluorescence measurements can be used to identify the peak observed expression level and to estimate the overall rate of fluorescence increase during the early-to-middle part of the experiment.
